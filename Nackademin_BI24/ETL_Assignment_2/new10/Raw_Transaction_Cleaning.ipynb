# Loan Account Transaction Cleaning
This notebook processes `Raw_Loan_Transaction.csv` to create a cleansed layer for Power bi.
- **Staging**: Inspect data, analyze missing values and distributions, identify empty columns, standardize formats, validate key columns
- **Cleansed**: Convert numeric columns, save to `cleansed/cleansed_loan_transactions.csv`.


## Staging: Imports and Data Loading
Load raw data and required libraries.

```python
# Import libraries for data processing and date handling
import pandas as pd

# Load raw data with semicolon delimiter
df = pd.read_csv('Raw_Loan_Transaction.csv', delimiter=';')
```

## Staging: Data Inspection
Inspect data types, missing values, unique values, distributions, and identify empty columns.

```python
# Display first 5 rows
print("First 5 rows of raw data:")
print(df.head())

# Display data types and non-null counts
print("\nData info:")
print(df.info())

# Display missing values per column
print("\nNumber of NaN values per column:")
print(df.isna().sum())
```

```python
# Analyze unique values and frequency distribution for each column
for col in df.columns:
    print(f"\nAnalyzing {col}")
    print("Unique values:", df[col].unique()[:10])  # First 10 unique values
    print("Most frequent values:", df[col].value_counts().head(10))  # Top 10 most frequent
    print("-" * 50)

# Identify empty columns (all NaN)
empty_columns = df.columns[df.isna().all()].tolist()
print("\nColumns that are completely empty:", empty_columns)
```

## Staging: Standardize Formats
Convert numeric and date columns to consistent formats.

```python
# Convert numeric columns with commas to float
numeric_columns = ['TransactionAmount', 'TransactionAmountSEK']
for col in numeric_columns:
    try:
        df[col] = df[col].astype(str).str.replace(',', '.').astype(float)
    except Exception as e:
        print(f"Error converting {col}: {e}")

# Verify updated data types
print("\nUpdated data types:")
print(df.dtypes)
```

# Cleansed Layer
Perform transformations to create a clean dataset for Power BI:
- Remove empty columns.
- Save clean data to `cleansed/cleansed_loan_transactions.csv`.

## Cleansed: Remove Empty Columns
Remove columns with only NaN values.

```python
# Drop empty columns
df = df.drop(columns=empty_columns, errors='ignore')

# Verify remaining columns
print("Remaining columns:", df.columns.tolist())
```

## Cleansed: Save Clean Data
Save cleansed data for analytical layer and Power BI.

```python
# Save cleansed data
df.to_csv('cleansed_loan_transactions.csv', index=False)
print("Saved cleansed data to cleansed_loan_transactions.csv.")
```
